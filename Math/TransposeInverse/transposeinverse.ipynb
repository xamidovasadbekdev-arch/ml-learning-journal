```python
# Transpose = flipping rows and columns
import numpy as np
A = np.array([[1, 2, 3],
              [4, 5, 6]])

print("A shape:", A.shape)
print("A.T shape:", A.T.shape)

print(A.T)
```

```python
# Inverse = finding a matrix that when multiplied with the original matrix gives the identity matrix
A = np.array([[4, 7],
              [2, 6]])

A_inv = np.linalg.inv(A)

print("Inverse of A:")
print(A_inv)
print("A x A-1  =")
print(( A @ A_inv).round(0)) 
```
