```python
#Hosila
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
y = x ** 2

plt.plot(x, y)
plt.title('f(x) = x^2')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

```python
# Derivative: x^2 = 2x

print(f"x = -3 derivative : {2 *(-3)}  (going down)")
print(f"x = 0 derivative : {2*0}  (going stable)")
print(f"x = 3 derivative : {2 * 3}  (going up)")
```

```python
# vizualization : slope of tangent lines at different points

fig, ax  = plt.subplots(1, 1, figsize=(8,5))
ax.plot(x, y, 'b-', linewidth=2)

for x_point in [-3, -1, 0, 1, 2, 3]:
    slope = 2 * x_point # derivative = 2x
    y_point = x_point ** 2
    x_tangent = np.linspace(x_point - 1, x_point + 1, 10)
    y_tangent = slope * (x_tangent - x_point) + y_point
    ax.plot(x_tangent, y_tangent, 'r-', alpha=0.7)
    ax.plot(x_point, y_point, 'ro', markersize = 8)
    ax.annotate(f'slope={slope}', {x_point, y_point + 1}, fontsize=10, ha='center')

ax.set_title('f(x) = x^2 slope of tangent lines(derivative)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True)
plt.show()
```
