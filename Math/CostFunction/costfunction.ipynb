```python
# real prices
import numpy as np

y_real = np.array([50000, 90000, 70000])

# model predictions (with wrong weights)
y_pred = np.array([45000, 100000, 60000])

# errors
difference = y_pred - y_real
print("Difference:", difference)

```

```python
# squared errors
diff_squared = difference ** 2
print("Differences squared:", diff_squared)

#MSE    
mse = np.mean(diff_squared)
print(f"\nMSE: {mse:,.0f}")
print("\nMse small = good model")
print("Mse large = bad model")
```

```python
# Let's compare with a better model

y_good = np.array([51000, 89000, 71000])

diff_good = y_good - y_real
mse_good = np.mean(diff_good ** 2)

print(f" Bad model MSE: {mse:,.0f}")
print(f"Good model MSE: {mse_good:,.0f}")
print(f"\nGood model {mse/mse_good:.0f} x better than the bad model")
```

```python
# visual = cost function landscape
import matplotlib.pyplot as plt

# simple example with 1 weight

w_values = np.linspace(-10, 10, 100)

# simple info: y = 3x, one dot x=2, y=6

x_data, y_data = 2, 6
cost_values = [(x_data * w - y_data) ** 2 for w in w_values]
plt.plot(w_values, cost_values)
plt.xlabel("Weight (w)")
plt.ylabel("Cost (MSE)")
plt.title("Cost Function Landscape")
plt.axvline(x=3, color='r', linestyle='--', label='Optimal Weight (w=3)')
plt.legend()
plt.grid(True)
plt.show()
```
