### Actually what happens under - model.fit()

```python
# 1 
# y = 2x+1
import numpy as np
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([3, 5, 7, 9, 11])

m = len(X)

X_b = np.c_[np.ones((m, 1)), X]  # add x0 = 1 to each instance
theta = np.linalg.inv(X.T@X)@X.T@y
```

```python
import matplotlib.pyplot as plt
np.random.seed(42)

X_raw = 2 * np.random.rand(100, 1)
y = 4 + 3 * X_raw + np.random.randn(100, 1) * 0.5

plt.scatter(X_raw, y, alpha = 0.5)
plt.title("Ma'lumotlar")
plt.show()
```

```python
# bias uchun 1-lar ustuni qoshamiz

X_b = np.c_[np.ones((100, 1)), X_raw]
print(X_b[:3]) # har bir qatorda oldin 1 keyin x

# normal equation

theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

print(f" theta_0 (bias) = {theta[0,0]:.4f} (haqiqiy: 4)")
print(f" theta_1 (weight) = {theta[1,0]:.4f} (haqiqiy: 3)")
```

```python
# chiziq chizamiz

x_line = np.array([[0], [2]])
x_line_b  = np.c_[np.ones((2, 1)), x_line]
y_line = x_line_b @ theta

plt.scatter(X_raw, y, alpha=0.5)
plt.plot(x_line, y_line, 'r-', linewidth=3)
plt.title(f'y = {theta[0,0]:.2f} + {theta[1,0]:.2f}x')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_raw, y)

print(f"Bizning math:  theta_0 = {theta[0,0]:.6f}, theta_1={theta[1,0]:6f}")

print(f" Sklearn : theta_0 = {model.intercept_[0]:.6f}, theta_1 = {model.coef_[0][0]:.6f}")
```

```python
# bashorat ham bir xil


x_new = 1.5

print(f"Bizning math: {(np.array([1, x_new]) @ theta)[0]:.4f}")
print(f" Sklearn : {model.predict([[x_new]])[0][0]:.4f}")
```
