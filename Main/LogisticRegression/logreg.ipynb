# LogisticRegression in code

```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# oddiy ma'lumot : yosh va kassallik (0/1)
X = np.array([20, 25, 30, 35, 40, 45, 50, 55, 60, 65, 70, 75, 80]).reshape(-1, 1)
y = np.array([0, 0, 0, 0, 0, 0, 1, 0, 1, 1, 1, 1, 1])

# Linear Regression bilan urinib ko'ramiz
model = LinearRegression()
model.fit(X, y)

#bashorat
X_test = np.linspace(10, 100, 100).reshape(-1, 1)
y_pred = model.predict(X_test)

plt.figure(figsize=(10, 5))
plt.scatter(X, y, color='red', s=100, zorder=5, label='Haqiqiy')
plt.plot(X_test, y_pred, color='blue', linewidth=2, label = "Linear Regression") 
plt.axhline(y=1, color='gray', linestyle='--', alpha=0.5)
plt.fill_between(X_test.flatten(), -0.5, 0, alpha=0.1, color='red', label='Nonsense zona (<0)')
plt.fill_between(X_test.flatten(), 1, 1.5, alpha=0.1, color='red', label =' Nonsense zona (>1)')
plt.xlabel('yosh')
plt.ylabel('bashorat')
plt.title('Linear Regression classification uchun ishlamaydi')
plt.legend()
plt.ylim(-0.5, 1.5)
plt.grid(True)
plt.show()
```

```python
print(f" 10 yoshli uchun bashorat : {model.predict([[10]])[0]:3f} <-- Manfiy, Ma'nosiz")
print(f" 95 yoshli uchun bashorat : {model.predict([[95]])[0]:.3f} <-- 1 dan katta! Ma'nosiz")
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
print(f"Sigmoid(-10): {sigmoid(-10):.6f} --> 0 ga juda yaqin")
print(f"Sigmoid(-5): {sigmoid(-5):.6f} --> 0 ga yaqin")
print(f"Sigmoid(-1): {sigmoid(-1):.6f} --> 0.5 dan pastroq")
print(f"Sigmoid(0): {sigmoid(0):.6f} --> 0.5 ga teng")
print(f"Sigmoid(1): {sigmoid(1):.6f} --> 0.5 dan balandroq")
print(f"Sigmoid(5): {sigmoid(5):.6f} --> 1 ga yaqin")
print(f"Sigmoid(10): {sigmoid(10):.6f} --> 1 ga juda yaqin")
```

```python
# S-shaklidagi egri chiziq
z = np.linspace(-10, 10, 100)
plt.figure(figsize=(10, 5))
plt.plot(z, sigmoid(z), 'b-', linewidth=3)
plt.axhline(y=0.5, color='r', linestyle='--', alpha=0.7, label='y=0.5 (chegaraviy nuqta)')
plt.axhline(y=0, color='gray', linestyle=':', alpha=0.5)
plt.axhline(y=1, color='gray', linestyle=':', alpha=0.5)

# CHANGED: axhline to axvline
plt.axvline(x=0, color='gray', linestyle=':', alpha=0.5) 

plt.scatter([0], [0.5], color='red', s=100, zorder=5)

# CHANGED: arrostyle to arrowstyle
plt.annotate('z=0 sigma=0.5', xy=(0,0.5), xytext=(2, 0.4), fontsize=12, 
             arrowprops=dict(arrowstyle='->'))

plt.title('Sigmoid function: sigma(z) = 1 / (1 + e^-z)')
plt.xlabel('z (linear regression chiqishi)')
plt.ylabel('sigma(z) (ehtimollik)')
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```
