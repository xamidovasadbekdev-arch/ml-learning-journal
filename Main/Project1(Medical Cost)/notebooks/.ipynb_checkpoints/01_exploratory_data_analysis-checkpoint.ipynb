# Lets see the connection of features with target

```python
num_cols = ['age', 'bmi', 'children']
cat_cols = df.select_dtypes(include="object").columns.to_list()
num_cols, cat_cols
```

### Numerical cols with target plotting

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 8))

for idx, c in enumerate(num_cols):
    axes[idx].scatter(df[c], df.charges, color="steelblue", edgecolor="red")
    axes[idx].set_xlabel(c)
    axes[idx].set_ylabel("Charges - Target")
plt.suptitle("The connection of numerical features with Target(charges)")
plt.tight_layout()
plt.show()
```

## Encoding

```python
# One-hot encode region and drop the first category (baseline)
region_dummies = pd.get_dummies(df['region'], prefix='region', drop_first=True, dtype=int)
```

```python
df['smoker'] = df['smoker'].map({'no': 0, 'yes': 1})
df['sex'] = df['sex'].map({'female': 0, 'male': 1})
```
